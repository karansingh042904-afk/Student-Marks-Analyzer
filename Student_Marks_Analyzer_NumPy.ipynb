# Student Marks Analyzer using NumPy

## Project overview
This beginner-friendly project generates marks for **50 students across five subjects** and analyzes performance using NumPy arrays, slicing, Boolean masks, averages, and grade bands.

- **Tools:** Python, NumPy, Matplotlib
- **Dataset:** Synthetic marks generated for demonstration, not real student records.
- **Goal:** Identify toppers, students needing support, subject-wise averages, grade bands, and pass percentage.

## 1. Import libraries
NumPy handles the arrays and calculations. Matplotlib creates a chart to make the results easier to understand.

```python
import numpy as np
import matplotlib.pyplot as plt

# Use a fixed seed so the same example results appear every time.
np.random.seed(42)
```

## 2. Create student names, subject names, and marks
`np.random.randint(35, 101, size=(50, 5))` creates a 50-row by 5-column array. The upper bound is exclusive, so 101 allows marks up to 100.

```python
students = np.array([f"Student_{i:02d}" for i in range(1, 51)])
subjects = np.array(["Maths", "Science", "English", "Computer", "History"])

# Each row is one student; each column is one subject.
marks = np.random.randint(35, 101, size=(50, 5))

print("Marks array shape:", marks.shape)
print("Number of dimensions:", marks.ndim)
print("Number of marks:", marks.size)
print("Data type:", marks.dtype)
print("\nFirst 5 students' marks:")
print(marks[:5])
```

## 3. Explore the array with indexing and slicing
Slicing lets us select rows or columns without writing loops.

```python
print("First 10 students:\n", marks[:10])
print("\nLast subject marks for the first 5 students:", marks[:5, -1])
print("\nMaths and Science marks for the first 5 students:\n", marks[:5, 0:2])
```

## 4. Calculate student totals, averages, and grades
For this project, a student passes if they score **at least 40 in every subject**. The grade bands below are project rules, not official college grading rules.

```python
student_totals = marks.sum(axis=1)
student_averages = marks.mean(axis=1)
subject_averages = marks.mean(axis=0)

# Grade bands based on each student's average.
grades = np.select(
    [student_averages >= 90,
     student_averages >= 75,
     student_averages >= 60,
     student_averages >= 40],
    ["A", "B", "C", "D"],
    default="F"
)

passed_all_subjects = np.all(marks >= 40, axis=1)
failed_students = students[~passed_all_subjects]
topper_index = np.argmax(student_averages)

print(f"Class average: {marks.mean():.2f}")
print(f"Top student: {students[topper_index]} (average {student_averages[topper_index]:.2f})")
print(f"Students passing every subject: {passed_all_subjects.sum()} / {len(students)}")
print(f"Pass percentage: {passed_all_subjects.mean() * 100:.2f}%")
print("Students needing support:", failed_students if len(failed_students) else "None")
print("\nSubject-wise averages:")
for subject, avg in zip(subjects, subject_averages):
    print(f"{subject}: {avg:.2f}")
```

## 5. Use Boolean masks to find high and low scores
A Boolean mask is a True/False condition applied to an array. Here, we find students with any mark below 40 and students with at least one mark above 90.

```python
has_low_mark = np.any(marks < 40, axis=1)
has_high_mark = np.any(marks > 90, axis=1)

print("Students with at least one mark below 40:", students[has_low_mark])
print("Students with at least one mark above 90:", students[has_high_mark])

# Count students in each average-grade band.
for grade in ["A", "B", "C", "D", "F"]:
    print(f"Grade {grade}: {(grades == grade).sum()} student(s)")
```

## 6. Subject-wise toppers
Find the highest mark in each subject and show all students who achieved it.

```python
for col, subject in enumerate(subjects):
    highest = marks[:, col].max()
    winners = students[marks[:, col] == highest]
    print(f"{subject}: {highest} mark(s) - {', '.join(winners)}")
```

## 7. Visualize subject averages

```python
plt.figure(figsize=(8, 4))
plt.bar(subjects, subject_averages)
plt.ylim(0, 100)
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average mark")
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

## 8. Summary report

```python
print("STUDENT MARKS ANALYZER - SUMMARY")
print("-" * 38)
print(f"Students analyzed: {len(students)}")
print(f"Subjects: {len(subjects)}")
print(f"Overall average: {marks.mean():.2f}/100")
print(f"Highest student average: {student_averages.max():.2f}/100")
print(f"Students passing every subject: {passed_all_subjects.sum()}")
print(f"Students with at least one mark below 40: {has_low_mark.sum()}")
print(f"Pass percentage: {passed_all_subjects.mean() * 100:.2f}%")
```

## 9. Conclusion
This project demonstrates how NumPy can store and analyze a two-dimensional dataset efficiently. It uses array shape and data types, slicing, negative indexing, Boolean masks, aggregation (`sum`, `mean`, `max`), and `argmax` to summarize student performance.

### Limitations and possible improvements

- Grade thresholds are illustrative and should be changed to match the institution's official rules.
- A future version could read marks from a CSV file and export the report.
